#### el problema
 * un  Algoritmo que lea N números y determine el mayor y el menor de la serie en una sola pasada.

```python
# --- PROGRAMA: ENCONTRAR EL MAYOR Y EL MENOR DE N NÚMEROS ---

print("==================================================")
print(" 🤖 BIENVENIDO AL SIMULADOR IA: DETECTOR DE EXTREMOS ")
print("==================================================")
print("Este sistema analizará números en tiempo real")
print("y determinará el mayor y el menor en una sola pasada.\n")

# Solicitamos la cantidad de números a analizar (N)
N = int(input("¿Cuántos números deseas ingresar en la serie?: "))

# Registramos el primer número para inicializar los extremos
print("\n--- Iniciando captura de datos ---")
numero = float(input("Ingresa el número 1: "))

# En la primera pasada, el primer número es tanto el mayor como el menor
mayor = numero
menor = numero

# Iniciamos un contador en 2 porque ya capturamos el primer número
contador = 2

# Ciclo para leer el resto de los números (hasta llegar a N)
while contador <= N:
    numero = float(input(f"Ingresa el número {contador}: "))

    # [Simulación de IA en una sola pasada]:
    # Evaluamos el número inmediatamente después de ser leído
    if numero > mayor:
        mayor = numero
    if numero < menor:
        menor = numero

    contador = contador + 1

# --- DESPLIEGUE DE RESULTADOS ---
print("\n==================================================")
print("         📊 ANALÍTICA FINAL DEL SISTEMA         ")
print("==================================================")
print(f" Total de números procesados: {N}")
print(f" 🔺 El número MAYOR detectado es: {mayor}")
print(f" 🔻 El número MENOR detectado es: {menor}")
print("==================================================")
```
